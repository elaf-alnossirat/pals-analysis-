```python
import pandas as pd

df = pd.read_csv("combat-attribute.csv", header=1)
```

```python
def clean_column(name):
    return (
        str(name).strip()
        .lower()
        .replace(" ", "_")
        .replace("-", "_")
        .replace("(", "")
        .replace(")", "")
    )

df.columns = [clean_column(col) for col in df.columns]
```

```python
df = df.dropna(axis=1, how='all')
```

```python
def clean_boolean_column(col):
    return col.replace({
        True: 1,
        False: 0,
        "TRUE": 1, "FALSE": 0,
        "true": 1, "false": 0,
        "Yes": 1, "No": 0,
        "YES": 1, "NO": 0,
        "yes": 1, "no": 0,
        "1": 1, "0": 0,
        "": 0,
        pd.NA: 0,
        None: 0
    }).fillna(0).astype(int)

# Liste des colonnes booléennes
bool_cols = ["ispal", "weaponequip", "nocturnal"]

# Application sur chaque colonne
for col in bool_cols:
    df[col] = clean_boolean_column(df[col])
```

```python
import pandas as pd

# 1. Charger le fichier proprement
df = pd.read_csv("job-skill.csv", header=0, skiprows=[0])

# 2. Nettoyer les noms de colonnes
def clean_column(name):
    return (
        str(name).strip()
        .lower()
        .replace(" ", "_")
        .replace("-", "_")
        .replace("(", "")
        .replace(")", "")
        .replace("=", "")
    )

df.columns = [clean_column(col) for col in df.columns]

# 3. Convertir les colonnes binaires (0/1)
binary_cols = [
    "make_a_fire", "watering", "planting", "generate_electricity", "manual", 
    "collection", "logging", "mining", "pharmaceutical", "cool_down", 
    "pasture", "carry"
]

for col in binary_cols:
    if col in df.columns:
        df[col] = df[col].replace({
            True: 1, False: 0, "yes": 1, "no": 0, "": 0, "1": 1, "0": 0
        }).fillna(0).astype(int)

# 4. Colonnes numériques (à adapter selon ton besoin)
numeric_cols = ["food_intake", "total_skills", "cool_down", "handling_speed", "pasture_minimum_output"]

for col in numeric_cols:
    if col in df.columns:
        df[col] = pd.to_numeric(df[col], errors="coerce").fillna(0).astype(int)

# 5. Exporter le fichier propre pour MySQL
df.to_csv("clean_job_skill_ready.csv", index=False)
```

```python
import pandas as pd

# 1. Charger le fichier avec les bonnes colonnes
df = pd.read_csv("hidden-attribute.csv")

# 2. Nettoyer les noms de colonnes
def clean_column(name):
    return (
        str(name).strip()
        .lower()
        .replace(" ", "_")
        .replace("-", "_")
        .replace("(", "")
        .replace(")", "")
    )

df.columns = [clean_column(col) for col in df.columns]

# 3. Convertir les booléens ('TRUE' / 'FALSE' / 'yes' / 'no') en 0/1
bool_cols = ["ispal"]
for col in bool_cols:
    df[col] = df[col].replace({
        "TRUE": 1, "FALSE": 0,
        "true": 1, "false": 0,
        "yes": 1, "no": 0,
        True: 1, False: 0,
        "": 0
    }).fillna(0).astype(int)

# 4. Convertir les colonnes numériques (à adapter selon ton besoin)
numeric_cols = ["cool_down", "carry", "pasture", "pharmaceutical", "mining"]
for col in numeric_cols:
    if col in df.columns:
        df[col] = pd.to_numeric(df[col], errors="coerce").fillna(0).astype(int)

# 5. Exporter le fichier nettoyé
df.to_csv("clean_hidden_attribute_ready.csv", index=False)
```

```python
import pandas as pd

# 1. Lire le fichier brut
df_raw = pd.read_csv("ordinary-boss-attribute.csv", header=None)

# 2. Sauter les 4 premières lignes explicatives
df = df_raw.iloc[4:, :]

# 3. Isoler les 3 blocs horizontaux
block1 = df.iloc[:, 0:3]
block2 = df.iloc[:, 3:6]
block3 = df.iloc[:, 6:8]  # Colonne 8 est la dernière utile

# 4. Renommer les colonnes de chaque bloc
block1.columns = ["name", "hp", "remote_attack"]
block2.columns = ["name", "hp", "remote_attack"]
block3.columns = ["name", "riding_speed"]

# 5. Ajouter les colonnes manquantes pour uniformiser
block1["riding_speed"] = None
block2["riding_speed"] = None
block3["remote_attack"] = None
block3["hp"] = None

# 6. Réorganiser les colonnes dans le même ordre
columns = ["name", "hp", "remote_attack", "riding_speed"]
block1 = block1[columns]
block2 = block2[columns]
block3 = block3[columns]

# 7. Fusionner les 3 blocs verticalement
df_final = pd.concat([block1, block2, block3], ignore_index=True)

# 8. Nettoyage des types (conversion en nombres là où possible)
df_final["hp"] = pd.to_numeric(df_final["hp"], errors="coerce")
df_final["remote_attack"] = pd.to_numeric(df_final["remote_attack"], errors="coerce")
df_final["riding_speed"] = pd.to_numeric(df_final["riding_speed"], errors="coerce")

# 9. Supprimer les lignes sans nom (vides)
df_final = df_final[df_final["name"].notna() & (df_final["name"] != "")]

# 10. Exporter en fichier prêt pour MySQL
df_final.to_csv("clean_ordinary_boss_attribute_ready.csv", index=False)
```

```python
import pandas as pd

# 1. Charger le fichier brut
df_raw = pd.read_csv("tower-boss-attribute.csv", header=None)

# 2. Extraire les noms des attributs (HP, melee, etc.)
attributes = df_raw.iloc[1:, 0].str.strip().str.lower().str.replace(" ", "_")
attributes = attributes.reset_index(drop=True)

# 3. Préparer une liste pour collecter les résultats
records = []

# 4. Traiter les colonnes 1 à la fin, par blocs de 2 (nom + stats)
for i in range(1, df_raw.shape[1], 2):
    boss_pair_name = df_raw.iloc[0, i]      # ex: "Victor & Griffin"
    boss_name = df_raw.iloc[0, i+1]         # ex: "Griffin"
    stats = df_raw.iloc[1:, i+1].reset_index(drop=True)

    # Créer un dictionnaire avec toutes les stats
    record = {"boss_pair": boss_pair_name, "boss_name": boss_name}
    for attr, val in zip(attributes, stats):
        try:
            record[attr] = int(val)
        except:
            record[attr] = None
    records.append(record)

# 5. Créer le DataFrame propre
df_boss = pd.DataFrame(records)

# 6. Exporter en CSV prêt pour MySQL
df_boss.to_csv("clean_tower_boss_attribute_ready.csv", index=False)
```

```python
# Jupyter Notebook : Nettoyage automatique des tables MySQL avec audit final

import mysql.connector

# Connexion MySQL
conn = mysql.connector.connect(
    host="localhost",
    user="root",
    password="root",
    database="palworld_database"
)
cursor = conn.cursor(dictionary=True)

# Fonction SELECT générique
def run_query(query):
    while cursor.nextset():
        cursor.fetchall()
    cursor.execute(query)
    return cursor.fetchall()

# Lister les tables
print("\u2705 Tables dans la base :")
tables = run_query("SHOW TABLES;")
for t in tables:
    print("-", list(t.values())[0])

# Analyse des valeurs manquantes
def check_nulls(table):
    print(f"\n\ud83d\udd0e Valeurs manquantes dans la table : {table}")
    cursor.execute(f"SELECT * FROM {table} LIMIT 1")
    cols = [desc[0] for desc in cursor.description]
    for col in cols:
        q = f"SELECT COUNT(*) AS null_count FROM {table} WHERE {col} IS NULL OR {col} = ''"
        result = run_query(q)
        if result[0]['null_count'] > 0:
            print(f"- {col} : {result[0]['null_count']} valeur(s) manquante(s)")

# Détection des doublons sur une colonne donnée
def check_duplicates(table, col):
    print(f"\n\ud83d\udcc9 Doublons dans {table} (colonne : {col})")
    q = f"""
    SELECT {col}, COUNT(*) AS cnt
    FROM {table}
    GROUP BY {col}
    HAVING cnt > 1
    """
    duplicates = run_query(q)
    for row in duplicates:
        print(f"- {row[col]} → {row['cnt']} doublons")

# Normalisation (minuscule) d'une colonne texte
def normalize_column(table, col):
    q = f"UPDATE {table} SET {col} = LOWER({col}) WHERE {col} IS NOT NULL"
    cursor.execute(q)
    conn.commit()
    print(f"\n✅ Colonne {col} de {table} normalisée.")

# Audit global des tables

def audit_table(table, key_column):
    print(f"\n\u2728 Audit de la table : {table}")
    check_nulls(table)
    check_duplicates(table, key_column)
    print("\u2705 Audit terminé.")

# Exemples d'audit pour plusieurs tables

# combat_attribute
audit_table("combat_attribute", "name")

# job_skill
audit_table("job_skill", "english_name")

# hidden_attribute
audit_table("hidden_attribute", "chinese_name")

# ordinary_boss_attribute
audit_table("ordinary_boss_attribute", "name")

# tower_boss_attribute
audit_table("tower_boss_attribute", "boss_name")

# refresh_area
audit_table("refresh_area", "name")

# Fermer la connexion
cursor.close()
conn.close()
```
